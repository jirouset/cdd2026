# Cofoldings Analysis: CXCR4 Ligand Contacts

```python
import pandas as pd

df_contacts = pd.read_csv("datasets/CXCR4_contacts.csv")
df_contacts.columns = [
    int(c) if c.lstrip("-").isdigit() else c for c in df_contacts.columns
]
df_labels = pd.read_csv("datasets/CXCR4_for_ml.csv")
print(f"Contact dataset: {df_contacts.shape}")
```

```python
# CIF availability report
# runs_processed == 0  -  all 5 CIF files were missing for that ligand
n_total_labelled = len(df_labels)
n_in_contacts = len(df_contacts[df_contacts["Inhibitor"].notna()])
n_no_cif = (df_contacts["runs_processed"] == 0).sum()
n_no_cif_labelled = (
    df_contacts[df_contacts["Inhibitor"].notna()]["runs_processed"] == 0
).sum()

print(f"Labelled ligands in CXCR4_for_ml.csv      : {n_total_labelled}")
print(f"Labelled ligands in contact dataset        : {n_in_contacts}")
print(f"  - no CIF found (runs_processed = 0)     : {n_no_cif_labelled}"
      f"  ({100 * n_no_cif_labelled / n_total_labelled:.1f}% of CXCR4_for_ml ligands)")
print(f"  - at least 1 CIF found                  : {n_in_contacts - n_no_cif_labelled}"
      f"  ({100 * (n_in_contacts - n_no_cif_labelled) / n_total_labelled:.1f}% of CXCR4_for_ml ligands)")
```

## Analysis: Actives vs Inactives in Boltz Cofoldings

Only ligands with a matched Inhibitor label **and** at least one successful run are used.

```python
from cofoldings_analysis import (
    prepare_valid_df,
    plot_contact_count_boxplot,
    plot_contact_count_boxplot_peptides,
    plot_contact_frequency_heatmap,
    run_mannwhitney_test,
    plot_residue_contact_diff,
    plot_confidence_boxplot,
    plot_confidence_boxplot_peptides,
)

df_valid, res_cols = prepare_valid_df(df_contacts)
```

```python
df_boltz = pd.read_csv("datasets/CXCR4_chembl_ligands_boltz.csv", sep=";")
```

### 0. Confidence scores - Actives vs Inactives

Boltz outputs four per-run confidence metrics. Values are averaged across all available runs per ligand before comparing groups.

```python
plot_confidence_boxplot(df_valid, confidence_dir="boltz_confidence/")
```

```python
plot_confidence_boxplot_peptides(df_valid, df_boltz, confidence_dir="boltz_confidence/")
```

### 1a. Boxplot - total contacts per ligand (actives vs inactives)

```python
plot_contact_count_boxplot(df_valid)
```

### 1b. Boxplot — peptide-based ligands highlighted

Peptide-based ligands (≥ 3 amide bonds) are shown as orange diamonds. They are expected to have higher mean contact counts because peptides are larger and interact with more residues.

```python
plot_contact_count_boxplot_peptides(df_valid, df_boltz)
```

On axis y is the average number of residues in contact of the 5 runs.

We can see that active tend to have more contacts in cofoldings with CXCR4 than inactive ligands. However, the median is lower for inhibitors..

Further analysis ideas:
* consider confidence - drop samples with low confidence
* divide to peptides vs small molecules --> more precise model
* highlight according to IC50 value

### 2. Heatmap - per-residue contact frequency (actives vs inactives)

Each cell shows the **mean fraction of runs** in which that CXCR4 residue was in contact with the ligand (0 = never, 1 = all 5 runs). Only residues with 20 % contact frequency in either group are shown.

```python
plot_contact_frequency_heatmap(df_valid, res_cols, min_freq=0.20)
```

### 3. Mann-Whitney U test - per-residue significance

For each CXCR4 residue, test whether the contact counts differ between actives and inactives. P-values are corrected with the Benjaminiâ€“Hochberg FDR procedure.


For each residue independently, it asks: are the contact counts in the active group drawn from the same distribution as in the inactive group, or is one
  group systematically higher?

```python
df_mw = run_mannwhitney_test(df_valid, res_cols)
display(df_mw[df_mw["p_value"] < 0.05].head(5))
```

```python
plot_residue_contact_diff(df_mw, p_threshold=0.05)
```

## Machine Learning - Classify Actives vs Inactives from Contact Fingerprint

The feature matrix is the per-residue contact count vector for each ligand (number of Boltz runs, out of 5, in which each CXCR4 residue was in contact).  
Target: `Inhibitor` (1 = active, 0 = inactive).  
80/20 stratified train/test split.

```python
from train_model import prepare_data, train_RF, train_KNN, train_logistic_regression, train_ridge_regression, evaluate_model

train_X, train_y, test_X, test_y = prepare_data(df_valid, target_col="Inhibitor", feature_cols=res_cols)
```

### Random Forest

```python
rf_model = train_RF(train_X, train_y)
evaluate_model(rf_model, test_X, test_y)
```

### K-Nearest Neighbours

```python
knn_model = train_KNN(train_X, train_y)
evaluate_model(knn_model, test_X, test_y)
```

### Logistic Regression (ElasticNet)

```python
lr_model = train_logistic_regression(train_X, train_y)
evaluate_model(lr_model, test_X, test_y)
```

### Ridge Regression (ElasticNet with l1_ratio=0)

```python
ridge_model = train_ridge_regression(train_X, train_y)
evaluate_model(ridge_model, test_X, test_y)
```

### XGBoost

```python
from train_model import train_XGB

xgb_model = train_XGB(train_X, train_y)
evaluate_model(xgb_model, test_X, test_y)
```

### ROC AUC & Feature Importance

```python
from train_model import plot_roc_auc, plot_feature_importance

models = {
    "RF": rf_model,
    "KNN": knn_model,
    # "LogReg": lr_model,
    # "Ridge": ridge_model,
    "XGB": xgb_model,
}
plot_roc_auc(models, test_X, test_y)
```

```python
plot_feature_importance(rf_model, res_cols)
plot_feature_importance(xgb_model, res_cols)
plot_feature_importance(lr_model, res_cols)
```
